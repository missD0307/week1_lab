## 1. Variables and types

A variable is a name for a value. Python works out the type for you.

````python
name = "Di"          # str   - text
age = 19                # int   - whole number
gpa = 9.99              # float - decimal number
is_student = True       # bool  - True or False

print(name)
print(age)
print(gpa)
print(is_student)

print(type(name), type(age), type(gpa), type(is_student))
````

### Printing with f-strings

Put `f` before the quotes, then write variables inside `{ }`.

````python
name = "Aziza"
age = 20
gpa = 3.75

print(f"{name} is {age} years old.")
print(f"Her GPA is {gpa}")
print(f"Rounded GPA: {gpa:.1f}")      # .1f means one decimal place
````

## 2. Numbers and arithmetic

| Symbol | Meaning | Example |
|---|---|---|
| `+` | add | `10 + 3` is `13` |
| `-` | subtract | `10 - 3` is `7` |
| `*` | multiply | `10 * 3` is `30` |
| `/` | divide | `10 / 3` is `3.333...` |
| `//` | divide, whole number only | `10 // 3` is `3` |
| `%` | remainder | `10 % 3` is `1` |
| `**` | power | `10 ** 3` is `1000` |

Note that `/` always gives a decimal, even for `10 / 5`.

````python
a = 10
b = 3

print("add      ", a + b)
print("subtract ", a - b)
print("multiply ", a * b)
print("divide   ", a / b)
print("whole    ", a // b)
print("remainder", a % b)
print("power    ", a ** b)

print()
print("rounded divide:", round(a / b, 2))
````

## 3. Strings

A string is text. Counting starts at **0**.

````python
course = "Machine Learning"

print("the string   ", course)
print("length       ", len(course))
print("first letter ", course[0])
print("last letter  ", course[-1])
print("first 7      ", course[0:7])
print("upper case   ", course.upper())
print("lower case   ", course.lower())
print("count of 'n' ", course.count("n"))
print("replaced     ", course.replace(" ", "_"))
````

## 4. Lists

A list holds several values in order. Counting starts at 0.

````python
marks = [72, 65, 90, 48, 83]

print("the list ", marks)
print("length   ", len(marks))
print("first    ", marks[0])
print("last     ", marks[-1])
print("total    ", sum(marks))
print("average  ", sum(marks) / len(marks))
print("highest  ", max(marks))
print("lowest   ", min(marks))
print("sorted   ", sorted(marks))
````

````python
marks = [72, 65, 90]

marks.append(48)          # add to the end
print("after append", marks)

marks.remove(65)          # remove a value
print("after remove", marks)

marks[0] = 100            # change one item
print("after change", marks)
````

## 5. Dictionaries

A dictionary stores pairs: a **key** and its **value**. You look things up by key
instead of by position.

````python
student = {"name": "Bekzod", "age": 21, "course": "AIML202"}

print(student)
print("name is", student["name"])

student["mark"] = 78          # add a new pair
student["age"] = 22           # change an existing one

print()
print(student)
print("keys  ", list(student.keys()))
print("values", list(student.values()))

print()
for key in student:
    print(key, "=", student[key])
````

## 6. Making decisions: if / elif / else

The indented lines under `if` only run when the condition is true.

Comparisons: `==` equal, `!=` not equal, `<` `>` `<=` `>=`.

````python
mark = 76

if mark >= 85:
    print("Grade A")
elif mark >= 70:
    print("Grade B")
elif mark >= 55:
    print("Grade C")
elif mark >= 40:
    print("Grade D")
else:
    print("Grade F")

print()
if mark >= 40 and mark <= 100:
    print("Valid passing mark")
````

## 7. Repeating: the for loop

A `for` loop repeats once for each item.

`range(1, 6)` gives 1, 2, 3, 4, 5 — the second number is **not** included.

````python
marks = [72, 65, 90]

for m in marks:
    print("mark:", m)

print()
for i in range(1, 6):
    print(i, end=" ")

print()
print()
total = 0
for m in marks:
    total = total + m
print("total is", total)
````

## 8. Repeating: the while loop

A `while` loop keeps going as long as its condition stays true. Make sure something
inside the loop eventually makes it false, or it never stops.

````python
count = 5
while count > 0:
    print(count, end=" ")
    count = count - 1
print("Go!")

print()
money = 100
while money >= 30:
    money = money - 30
    print("spent 30, left:", money)
````

## 9. Functions

A function is a piece of code with a name, so you can use it many times.
`return` sends a value back to whoever called it.

````python
def greet(name):
    return f"Hello, {name}!"


def area(width, height):
    return width * height


def average(numbers):
    return sum(numbers) / len(numbers)


print(greet("Aziza"))
print(area(5, 3))
print(average([72, 65, 90]))
print(round(average([72, 65, 90]), 1))
````

## 10. Putting it together

A small program using everything above.

````python
gradebook = {"Aziza": 88, "Bekzod": 64, "Dilnoza": 95, "Eldor": 42}


def grade(mark):
    if mark >= 85:
        return "A"
    elif mark >= 70:
        return "B"
    elif mark >= 55:
        return "C"
    else:
        return "F"


print("Name       Mark  Grade")
for name in gradebook:
    mark = gradebook[name]
    print(f"{name:<10} {mark:>4}  {grade(mark)}")

total = sum(gradebook.values())
print()
print("Class average:", round(total / len(gradebook), 1))
````

## Exercise 10 — Small program  (10 marks)

Start with this dictionary:

```python
gradebook = {"Aziza": 88, "Bekzod": 64, "Dilnoza": 95, "Eldor": 42, "Farrukh": 73}
```

Write a program that prints:

1. every student and their mark, one per line; *(3 marks)*
2. the class average, rounded to one decimal place; *(3 marks)*
3. how many students scored 60 or more; *(2 marks)*
4. the name of the student with the highest mark. *(2 marks)*

**Expected output**

```
Aziza      88
Bekzod     64
Dilnoza    95
Eldor      42
Farrukh    73

class average: 72.4
scored 60 or more: 4
top student: Dilnoza with 95
```

````python
# Exercise 10 - your code here
gradebook = {"Aziza": 88, "Bekzod": 64, "Dilnoza": 95, "Eldor": 42, "Farrukh": 73}

def grade(mark):
  if mark >= 85:
    return "A"
  elif mark >= 70:
    return "B"
  elif mark >= 55:
    return "C"
  else:
    return "F"

for name in gradebook:
    mark = gradebook[name]
    print(f"{name:<10} {mark:>4}")

total = sum(gradebook.values())
print()
print("class average:", round(total / len(gradebook), 1))

scored_60_or_more = sum(1 for mark in gradebook.values() if mark >= 60)
print(f"scored 60 or more: {scored_60_or_more}")
top_student = max(gradebook, key=gradebook.get)
print(f"top student: {top_student} with {gradebook[top_student]}")
````
